## 리스트(list)

````python
# 리스트는 여러 값을 순서대로 담는, 가장 기본적이며 가장 자주 쓰이는 자료 구조이다. 대괄호 [ ] 로 만든다.
scores = [80, 90, 75, 100, 65]
print(scores)
print(len(scores))      # 길이
print(scores[0])        # 첫 번째 (인덱스는 0부터)
print(scores[-1])       # 마지막
print(scores[1:4])      # 슬라이싱: 1,2,3번째
[80, 90, 75, 100, 65]
5
80
65
[90, 75, 100]
````

````python
# 리스트는 "수정 가능(mutable)" 하다. 값을 추가·삭제·변경할 수 있다.
scores.append(88)        # 끝에 추가
print(scores)
scores[0] = 95           # 값 변경
print(scores)
scores.remove(75)        # 값으로 삭제
print(scores)
scores.sort()            # 정렬
print(scores)
[80, 90, 75, 100, 65, 88]
[95, 90, 75, 100, 65, 88]
[95, 90, 100, 65, 88]
[65, 88, 90, 95, 100]
````

## 튜플(tuple)

````python
# 튜플은 리스트와 비슷하지만 "수정 불가(immutable)" 한 자료 구조다. 소괄호 ( ) 또는 쉼표로 만든다. 함수가 여러 값을 반환할 때 자주 사용된다.
pt = (3, 4)
x, y = pt           # 분해 대입
print(x, y)
# 두 값 바꿔치기
a, b = 1, 2
a, b = b, a
print(a, b)
````

## 딕셔너리(dict)

````python
# 딕셔너리는 키(key)와 값(value)을 짝지어 저장하는 자료 구조이다. 중괄호 { } 와 콜론(:)을 쓴다.
person = {"name": "Hong", "age": 30, "city": "Seoul"}
print(person["name"])              # 키로 값 꺼내기
person["age"] = 31                 # 값 수정
person["job"] = "Researcher"       # 새 항목 추가
print(person)
print("name" in person)            # 키 존재 여부
````

## 집합(set)

````python
# 집합은 중복을 허용하지 않는, 순서가 없는 묶음이다. 중괄호 { } 또는 set() 로 만든다. 합집합·교집합·차집합 연산을 직관적으로 할 수 있다.
A = {1, 2, 3, 3}
B = {2, 3, 4}
print(A)            # 중복은 자동 제거: {1, 2, 3}
print(A | B)        # 합집합
print(A & B)        # 교집합
print(A - B)        # 차집합
{1, 2, 3}
{1, 2, 3, 4}
{2, 3}
{1}
````

## 조건문 — if / elif / else

````python
# 파이썬은 들여쓰기로 "이 코드가 어디에 속해 있는지"를 표현한다. 본문에서는 들여쓰기를 보통 공백 4칸으로 한다.
x = 75
if x >= 90:
    grade = "A"
elif x >= 80:
    grade = "B"
elif x >= 70:
    grade = "C"
else:
    grade = "D"
print(grade)
````

## 반복문 — for, while

````python
# 리스트나 다른 순차형 자료를 차례로 훑어볼 때는 for 문을 쓴다. 반복 횟수를 직접 정하려면 range() 와 함께 쓴다.
scores = [80, 90, 75, 100, 65]
for s in scores:
    print(s)
print("---")
for i in range(3):           # 0, 1, 2
    print("반복", i)
80
90
75
100
65
````

````python
# "인덱스와 값을 함께" 보고 싶다면 enumerate() 가 깔끔하다.
names = ["Alice", "Bob", "Charlie"]
for i, name in enumerate(names):
    print(i, name)
````

## 리스트 컴프리헨션

````python
# "리스트의 각 원소에 어떤 연산을 적용해 새 리스트를 만들기"는 다음과 같이 한 줄로 쓸 수 있다.
scores = [80, 90, 75, 100, 65]
doubled = [s * 2 for s in scores]
high = [s for s in scores if s >= 85]
print(doubled)
print(high)
[160, 180, 150, 200, 130]
[90, 100]
````

## while 문

````python
# 정해진 횟수가 아니라 "조건이 참인 동안" 반복하려면 while 문을 쓴다.
n = 1
while n < 100:
    n = n * 2
print(n)
128
````

## 함수 정의 — def

````python
# 자주 쓰는 작업은 함수로 묶어 두면 같은 코드를 다시 쓸 필요가 없다.
def average(xs):
    return sum(xs) / len(xs)
def grade(x, cutoffs=(90, 80, 70)):
    if x >= cutoffs[0]:
        return "A"
    elif x >= cutoffs[1]:
        return "B"
    elif x >= cutoffs[2]:
        return "C"
    else:
        return "D"
print(average([80, 90, 75, 100, 65]))
print(grade(85))
82.0
B
````

## 난수 — np.random

## CSV 파일 읽기

````python
# 실제 자료는 보통 CSV(.csv) 또는 Excel(.xlsx) 파일로 주어진다. pandas는 한 줄로 이를 DataFrame 으로 불러온다.
# 인터넷의 CSV
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
from sklearn.datasets import load_iris
_iris = load_iris(as_frame=True)
iris = _iris.frame.rename(columns={"target": "species"})
iris["species"] = iris["species"].map(dict(enumerate(_iris.target_names)))
iris = iris.rename(columns={"sepal length (cm)": "sepal_length", "sepal width (cm)": "sepal_width", "petal length (cm)": "petal_length", "petal width (cm)": "petal_width"})
df = iris.copy()
# 또는 내 컴퓨터의 파일
# df = pd.read_csv("data/iris.csv")
# 한글 헤더가 깨지면 인코딩을 지정한다 (utf-8 또는 utf-8-sig)
# df = pd.read_csv("데이터/movie_data.csv", encoding="utf-8")
print(df.head())            # 처음 5행
print(df.shape)             # (행 수, 열 수)
print(df.columns.tolist())  # 열 이름들
print(df.dtypes)            # 열별 자료형
print(df.describe())        # 수치 열 요약 통계
````

## 행과 열 선택하기

## 정규분포 다루기

````python
# 정규분포 다루기
from scipy import stats
# 표준정규에서 확률밀도, 누적확률, 분위수
print(stats.norm.pdf(0))             # f(0)
print(stats.norm.cdf(1.96))          # P(Z <= 1.96)
print(stats.norm.ppf(0.975))         # 97.5% 분위수 ≈ 1.96
# 정규분포에서 표본 100개 뽑기
np.random.seed(0)
sample = stats.norm.rvs(loc=170, scale=10, size=100)
print(sample[:5].round(2))
0.3989422804014327
0.9750021048517795
1.959963984540054
````

## 선형회귀

````python
# 선형회귀
import statsmodels.formula.api as smf
# 가짜 데이터
rng = np.random.default_rng(0)
n = 100
x = rng.normal(size=n)
y = 2 + 3 * x + rng.normal(scale=1.0, size=n)
df = pd.DataFrame({"x": x, "y": y})
# y = b0 + b1*x
model = smf.ols("y ~ x", data=df).fit()
print(model.summary().tables[1])
print("R^2 =", model.rsquared.round(3))
````

````python
# 두 개 이상의 설명변수를 쓸 때는 + 로 잇고, 범주형 변수는 C(...)로 감싸 더미 변수를 자동 생성하게 한다.
# 다중회귀 + 범주형
df["g"] = np.where(df["x"] > 0, "A", "B")
m = smf.ols("y ~ x + C(g)", data=df).fit()
print(m.params)
````

## scikit-learn — 분류와 군집화

````python
# scikit-learn("사이킷런", import 시에는 sklearn)은 머신러닝 알고리즘 표준 라이브러리이다. 모든 모형이 "fit → predict"라는 동일한 사용법을 따르는 것이 큰 장점이다.
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.datasets import load_iris
_iris = load_iris(as_frame=True)
iris = _iris.frame.rename(columns={"target": "species"})
iris["species"] = iris["species"].map(dict(enumerate(_iris.target_names)))
iris = iris.rename(columns={"sepal length (cm)": "sepal_length", "sepal width (cm)": "sepal_width", "petal length (cm)": "petal_length", "petal width (cm)": "petal_width"})
df = iris.copy()
X = iris.drop(columns=["species"])
y = iris["species"]
# 1) 학습용·검증용 나누기
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.3, random_state=0, stratify=y)
# 2) 표준화
scaler = StandardScaler()
X_tr_s = scaler.fit_transform(X_tr)
X_te_s = scaler.transform(X_te)
# 3) 모형 적합 (로지스틱 회귀 분류기)
clf = LogisticRegression(max_iter=1000)
clf.fit(X_tr_s, y_tr)
# 4) 예측·평가
y_hat = clf.predict(X_te_s)
print("정확도:", accuracy_score(y_te, y_hat))
print(confusion_matrix(y_te, y_hat))
정확도: 0.9111111111111111
````

## A.1.2 자료 변환: Wide ↔ Long

````python
# [파이썬 코드 1-4]
# 치료제 데이터 (wide 형식)
trt_wide = pd.DataFrame({
    "id": [1, 2, 3],
    "a":  [23, 17, 9],
    "b":  [16, None, 11],
})
trt_wide
````

````python
# [파이썬 코드 1-5]
# wide → long
trt_long = trt_wide.melt(
    id_vars="id",                   # 그대로 유지할 열
    value_vars=["a", "b"],          # 하나로 모을 열
    var_name="treatment",           # 열 이름이 담길 새 변수
    value_name="response",          # 값이 담길 새 변수
)
trt_long
````

````python
# [파이썬 코드 1-6]
# long → wide
trt_wide_recovered = (
    trt_long
    .pivot_table(index="id", columns="treatment", values="response")
    .reset_index()
)
trt_wide_recovered
````

## A.1.3 데이터 변환의 5가지 문법: 배급사 RTP 분석

````python
# [파이썬 코드 1-8]
df = 영화데이터.copy()
df["대표배급사"] = (
    df["배급사"]
    .astype("string")
    .str.split(",")
    .str[0]
    .str.strip()
)
````

````python
# [파이썬 코드 1-9]
base = df.loc[:, ["대표배급사", "개봉년도", "관객수", "영화명"]]
````

````python
# [파이썬 코드 1-10]
base = base.dropna(subset=["대표배급사", "개봉년도", "관객수"])
````

````python
# [파이썬 코드 1-11]
stats = (
    base
    .groupby(["대표배급사", "개봉년도"], as_index=False)
    .agg(
        평균관객수=("관객수", "mean"),
        영화수=("영화명", "count"),
    )
)
````

````python
# [파이썬 코드 1-12]
peak = (
    stats
    .groupby("대표배급사", as_index=False)
    .agg(peak_평균관객수=("평균관객수", "max"))
)
stats = stats.merge(peak, on="대표배급사", how="left")
stats["RTP"] = stats["평균관객수"] / stats["peak_평균관객수"]
````

````python
# [파이썬 코드 1-13]
recent_year = int(base["개봉년도"].max())
recent = (
    stats
    .query("개봉년도 == @recent_year")
    .sort_values("RTP")
    .loc[:, ["대표배급사", "개봉년도", "영화수",
             "평균관객수", "peak_평균관객수", "RTP"]]
    .head(10)
)
recent
````

## A.1.4 자료 결합

````python
# [파이썬 코드 1-15]
# 자치구별 문화행사 수 요약
문화_요약 = (
    서울시문화행사정보
    .groupby("자치구", dropna=False)
    .size()
    .reset_index(name="행사수")
)
# 자치구별 아파트 거래 요약
부동산_요약 = (
    아파트데이터
    .query("건축년도 != 0")
    .groupby("자치구명", dropna=False)
    .agg(
        거래건수=("가격", "count"),
        물건금액_평균=("가격", "mean"),
        물건금액_표준편차=("가격", "std"),
        건물면적_평균=("건물면적", "mean"),
        건물면적_표준편차=("건물면적", "std"),
    )
    .reset_index()
)
````

````python
# [파이썬 코드 1-16]
# inner join: 두 표 모두에 존재하는 자치구만
문화_부동산_inner = pd.merge(
    문화_요약, 부동산_요약,
    how="inner", left_on="자치구", right_on="자치구명",
)
print(f"inner join 행 수: {len(문화_부동산_inner)}")
# left join: 문화행사 표 기준, 아파트 정보가 없으면 결측값
문화_부동산_left = pd.merge(
    문화_요약, 부동산_요약,
    how="left", left_on="자치구", right_on="자치구명",
)
print(f"left join 행 수: {len(문화_부동산_left)}")
# full join: 양쪽 표의 모든 자치구 포함
문화_부동산_full = pd.merge(
    문화_요약, 부동산_요약,
    how="outer", left_on="자치구", right_on="자치구명",
)
print(f"full join 행 수: {len(문화_부동산_full)}")
````

## 커널 밀도 추정(KDE)과 대역폭 비교

````python
# [파이썬 코드 1-21]
from scipy.stats import gaussian_kde
# KDE를 그릴 x축 그리드 생성
x_grid = np.linspace(x.min() - 0.2, x.max() + 0.2, 1000)
# 대역폭별 KDE 계산
bw_list = [0.05, 0.1, 0.2]
density_list = []
for bw in bw_list:
    kde = gaussian_kde(x, bw_method=bw / np.std(x, ddof=1))
    y = kde(x_grid)
    density_list.append(pd.DataFrame({
        "x": x_grid, "y": y,
        "bandwidth": f"bandwidth = {bw}",
    }))
density_df = pd.concat(density_list, ignore_index=True)
````

## A.2.2 선형회귀분석 실습

````python
# [파이썬 코드 2-5]
# 선형회귀분석을 위한 모듈 로드
import statsmodels.formula.api as smf
# 다중선형회귀 모형의 형태 선언 (로그 연봉 ~ 경기수 + 평균 자책점)
formula_G_ERA = "log_salary ~ G + ERA"
# 다중선형회귀모형 적합
fit_G_ERA = smf.ols(formula=formula_G_ERA, data=baseball).fit()
# 다중선형회귀모형 적합 결과 확인
print(fit_G_ERA.summary())
````

````python
# [파이썬 코드 2-6]
# 선형모형을 이용한 예측값 계산
fitted_vals = fit_G_ERA.fittedvalues
# 스튜던트화 잔차 계산
influence = fit_G_ERA.get_influence()
residuals = influence.resid_studentized_internal
````

````python
# [파이썬 코드 2-7]
# 잔차도 그리기
plt.figure()
sns.residplot(x=fitted_vals, y=residuals)
plt.axhline(0, color='gray', linestyle='--')
plt.show()
````

````python
# [파이썬 코드 2-8]
from statsmodels.graphics.gofplots import qqplot
# 정규분포 Q-Q 플롯 그리기
qqplot(residuals, line='45', fit = True, alpha = 0.5)
plt.show()
````

````python
# [파이썬 코드 2-11]
# 교호작용을 고려하지 않은 모형
factor_fit1 = smf.ols("log_salary ~ G + closer", data = baseball).fit()
print(factor_fit1.summary())
````

````python
# [파이썬 코드 2-12]
# 교호작용을 고려한 모형
factor_fit2 = smf.ols("log_salary ~ G * closer", data = baseball).fit()
print(factor_fit2.summary())
````

````python
# [파이썬 코드 2-13]
# 시각화를 위한 설명변수의 range에 해당하는 x값들을 구하기, 길이 100
var_range = np.linspace(baseball['G'].min(), baseball['G'].max(), 100)
# 적합한 값인 fitted value를 담아줄 데이터프레임 만들어서 넣기
pred_df1 = pd.concat([
    pd.DataFrame({'G': var_range, 'closer': 0}),
    pd.DataFrame({'G': var_range, 'closer': 1})
], ignore_index=True)
pred_df1['closer'] = pred_df1['closer'].astype('category')
pred_df1['log_salary'] = factor_fit1.predict(pred_df1)
# 그래프 그리기
plt.figure()
for label_val, color in zip([0, 1], ['red', 'blue']):
    subset_data = baseball[baseball['closer'] == label_val]
    subset_pred = pred_df1[pred_df1['closer'] == label_val]
    plt.scatter(subset_data['G'], subset_data['log_salary'], color=color,
                label='relief pitcher' if label_val == 0 else 'closer')
    plt.plot(subset_pred['G'], subset_pred['log_salary'], color=color)
plt.legend(loc='lower right')
plt.show()
````

## 커널 회귀분석

````python
# [파이썬 코드 2-15]
from statsmodels.nonparametric.kernel_regression import KernelReg
# 커널 회귀모형 선언 및 적합
# endog: 반응변수, exog: 설명변수, var_type='c': 연속형 변수 의미, bw=[3]: 대역폭 설정
kr = KernelReg(endog=baseball.log_salary, exog=baseball.G, var_type='c', reg_type='lc', bw=[3])
# 적합된 값 계산
y_fit, _ = kr.fit()
````

## 스플라인 회귀

````python
# [파이썬 코드 2-17]
import statsmodels.api as sm
from statsmodels.gam.api import BSplines # B-스플라인 사용하기 위한 라이브러리
# 매듭점 설정: G의 1사분위수, 중앙값, 3사분위수를 기준으로 구간을 나눔
G_knots = baseball['G'].quantile([0.25, 0.50, 0.75]).to_list()
# B-스플라인 기저 생성
bs = BSplines(x=baseball[['G']], df=[7], degree=[3], knot_kwds=[{"knots": G_knots}])
# 반응변수와 설계행렬 생성
X_spline = sm.add_constant(bs.basis, has_constant='add')
y = baseball['log_salary']
# 회귀모형 적합
fit_spline = sm.GLM(y, X_spline, family=sm.families.Gaussian()).fit()
````

````python
# [파이썬 코드 2-18]
# 시각화를 위한 촘촘한 경기 수 값 생성
G_grid = np.linspace(baseball['G'].min(), baseball['G'].max(), 200)
G_grid_df = pd.DataFrame({'G': G_grid})
# 예측용 데이터에 대한 B-스플라인 기저 생성
bs_pred = BSplines(x=G_grid_df[['G']], df=[7], degree=[3],
          knot_kwds=[{"knots":G_knots}])
# 예측값 계산
X_pred = sm.add_constant(bs_pred.basis, has_constant='add')
y_spline_pred = fit_spline.predict(X_pred)
# 산점도와 스플라인 곡선 시각화
plt.figure()
plt.scatter(baseball['G'], baseball['log_salary'],color='gray')
plt.plot(G_grid, y_spline_pred, color='blue')
plt.show()
````

## 스무딩 스플라인

````python
# [파이썬 코드 2-19]
from scipy.interpolate import make_smoothing_spline
# 전처리: 동일한 G 값에 대해서는 log_salary의 평균을 사용
smooth_df = baseball[['G', 'log_salary']].groupby('G', as_index=False).mean()
# 전처리: 경기수 G 기준 오름차순 정렬
smooth_df = smooth_df.sort_values(by='G')
# 스무딩 스플라인 적합
# lam: 매끄러움의 정도를 조절하는 조정 모수 (lambda)
fit_smooth = make_smoothing_spline(smooth_df['G'], smooth_df['log_salary'], lam=182)
````

````python
# [파이썬 코드 2-20]
# 시각화를 위한 촘촘한 경기 수 G 생성
plot_x = np.linspace( smooth_df['G'].min(), smooth_df['G'].max(), 300)
# 적합된 모형을 사용하여 예측값 계산
plot_y = fit_smooth(plot_x)
# 결과 시각화
plt.figure()
plt.scatter(baseball['G'], baseball['log_salary'],color='gray')
plt.plot(plot_x, plot_y, color='blue')
plt.show()
````

````python
# [파이썬 코드 2-21]
from pygam import GAM, s
# GAM 모형 정의: 각 설명변수에 대해 3차 스플라인(spline_order=3) 적용
# n_splines: 매듭점의 개수와 유사한 개념으로, 곡선의 유연성을 결정
gam_fit = GAM( s(0, spline_order=3, n_splines=8)
      + s(1, spline_order=3, n_splines=8)).fit(baseball[['G', 'ERA']], baseball['log_salary'])
# 적합 결과 요약 확인
print(gam_fit.summary())
````

````python
# [파이썬 코드 2-22]
# G의 변화에 따른 부분효과의 변화를 관찰하기 위한 촘촘한 G의 값 생성
G_grid = np.linspace(baseball['G'].min(), baseball['G'].max(), 100)
# G 부분효과 계산을 위해 다른변수(ERA)를 평균값으로 고정하여 환경을 통제
ERA_mean = baseball['ERA'].mean()
X_G = np.column_stack([G_grid, np.full_like(G_grid, ERA_mean)])
# G의 부분효과 계산
G_partial = gam_fit.partial_dependence(term=0, X=X_G)
# --- 부분 잔차(Partial Residuals) 계산 ---
# 전체 예측값 계산
overall_predictions = gam_fit.predict(baseball[['G', 'ERA']])
# 전체 잔차(Full Residuals) 계산: 실제값 - 전체 예측값
full_residuals = baseball['log_salary'] - overall_predictions
# 부분 잔차 계산
G_resid = full_residuals + gam_fit.partial_dependence(term=0, X=baseball[['G', 'ERA']].values)
````

````python
# [파이썬 코드 2-23]
plt.figure()
# 부분 효과 곡선 (파란선)
plt.plot(G_grid, G_partial, color='blue')
# 부분 잔차 산점도 (회색점)
plt.scatter(baseball['G'], G_resid, color='grey')
plt.show()
````

## 분류 성능의 평가 — 혼동행렬, 민감도, 특이도

````python
# [파이썬 코드 3-16]
pred_prob = multi_fit.predict(data)
pred_class = (pred_prob > 0.5).astype(int)
````

````python
# [파이썬 코드 3-17]
from sklearn.metrics import confusion_matrix
conf_mat = pd.crosstab(pred_class, data["diabetes_int"],
                       rownames=["predicted"], colnames=["actual"])
conf_mat
````

````python
# [파이썬 코드 3-18]
TN = conf_mat.loc[0, 0]
FN = conf_mat.loc[0, 1]
FP = conf_mat.loc[1, 0]
TP = conf_mat.loc[1, 1]
accuracy    = (TP + TN) / conf_mat.values.sum()
sensitivity = TP / (TP + FN)
specificity = TN / (TN + FP)
print(f"정확도 (accuracy):    accuracy:.4f")
print(f"민감도 (sensitivity): sensitivity:.4f")
print(f"특이도 (specificity): specificity:.4f")
````

````python
# [파이썬 코드 3-19]
def calc_metrics(prob, actual, threshold):
    pred = (prob > threshold).astype(int)
    cm = pd.crosstab(pred, actual).reindex(index=[0, 1], columns=[0, 1], fill_value=0)
    TN, FN = cm.loc[0, 0], cm.loc[0, 1]
    FP, TP = cm.loc[1, 0], cm.loc[1, 1]
    return {
        "임계값":   threshold,
        "정확도":    (TP + TN) / cm.values.sum(),
        "sensitivity": TP / (TP + FN) if (TP + FN) else np.nan,
        "specificity": TN / (TN + FP) if (TN + FP) else np.nan,
    }
thresholds = [0.3, 0.4, 0.5, 0.6, 0.7]
metrics_df = pd.DataFrame([calc_metrics(pred_prob, data["diabetes_int"], t) for t in thresholds])
metrics_df.round(4)
````

````python
# [파이썬 코드 3-20]
thr_seq = np.arange(0.05, 0.96, 0.01)
trade_df = pd.DataFrame([calc_metrics(pred_prob, data["diabetes_int"], t) for t in thr_seq])
plt.figure(figsize=(8, 5))
plt.plot(trade_df["임계값"], trade_df["sensitivity"],
         color="#E07B91", linewidth=1.5, label="sensitivity")
plt.plot(trade_df["임계값"], trade_df["specificity"],
         color="#6888BE", linewidth=1.5, label="specificity")
plt.xlabel("임계값")
plt.ylabel("값")
plt.title("임계값에 따른 민감도와 특이도")
plt.legend()
plt.tight_layout()
plt.show()
````

## K-최근접 이웃(KNN) 분류기

````python
# [파이썬 코드 3-22]
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split, KFold
X = data[["age", "bmi"]].values
y = data["diabetes_int"].values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
````

````python
# [파이썬 코드 3-23]
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=2025
)
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
knn_pred = knn.predict(X_test)
print("Confusion matrix:")
print(pd.crosstab(knn_pred, y_test, rownames=["predicted"], colnames=["actual"]))
print(f"\n정확도: (knn_pred == y_test).mean():.4f")
````

````python
# [파이썬 코드 3-24]
k_vals = [1, 3, 5, 10, 20, 50, 100]
err_rates = []
for k in k_vals:
    pred = KNeighborsClassifier(n_neighbors=k).fit(X_train, y_train).predict(X_test)
    err_rates.append(np.mean(pred != y_test))
knn_results = pd.DataFrame({"K": k_vals, "error_rate": err_rates})
knn_results
````

````python
# [파이썬 코드 3-25]
plt.figure(figsize=(8, 5))
plt.plot(knn_results["K"], knn_results["error_rate"],
         color="#6888BE", marker="o", linewidth=1.5)
plt.xscale("log")
plt.xlabel("K (로그 척도)")
plt.ylabel("오류율")
plt.title("K에 따른 검정 오류율")
plt.tight_layout()
plt.show()
````

````python
# [파이썬 코드 3-26]
k_vals_cv = [1, 3, 5, 10, 15, 20, 30, 50, 100]
n_folds = 5
kf = KFold(n_splits=n_folds, shuffle=True, random_state=2025)
cv_err = []
for k in k_vals_cv:
    fold_err = []
    for train_idx, val_idx in kf.split(X_scaled):
        knn = KNeighborsClassifier(n_neighbors=k)
        knn.fit(X_scaled[train_idx], y[train_idx])
        pred = knn.predict(X_scaled[val_idx])
        fold_err.append(np.mean(pred != y[val_idx]))
    cv_err.append(np.mean(fold_err))
cv_results = pd.DataFrame({"K": k_vals_cv, "cv_error": cv_err})
cv_results
````

````python
# [파이썬 코드 3-27]
best_k = int(cv_results.loc[cv_results["cv_error"].idxmin(), "K"])
print("최적의 K:", best_k)
````

````python
# [파이썬 코드 3-31]
from sklearn.tree import DecisionTreeClassifier, plot_tree
# 범주형 변수를 더미로 변환하여 사용
X_tree = data[["age", "bmi", "hp"]].copy()
X_tree["hp"] = X_tree["hp"].astype(int)
y_tree = data["diabetes_int"].values
tree = DecisionTreeClassifier(max_depth=3, random_state=2025)
tree.fit(X_tree, y_tree)
print(f"훈련 정확도: tree.score(X_tree, y_tree):.4f")
````

````python
# [파이썬 코드 3-43]
import torch
import torch.nn as nn
class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, 16, kernel_size=3, padding=1), nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(16, 32, kernel_size=3, padding=1), nn.ReLU(),
            nn.MaxPool2d(2),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32 * 7 * 7, 64), nn.ReLU(),
            nn.Linear(64, num_classes),     # logits
        )
    def forward(self, x):
        return self.classifier(self.features(x))  # softmax는 손실함수에서 처리
````

````python
# [파이썬 코드 3-49]
def metrics(pred, actual):
    cm = pd.crosstab(pd.Series(pred), pd.Series(actual)).reindex(
        index=[0, 1], columns=[0, 1], fill_value=0)
    TN, FN = cm.loc[0, 0], cm.loc[0, 1]
    FP, TP = cm.loc[1, 0], cm.loc[1, 1]
    return {
        "정확도":    (TP + TN) / cm.values.sum(),
        "sensitivity": TP / (TP + FN) if (TP + FN) else np.nan,
        "specificity": TN / (TN + FP) if (TN + FP) else np.nan,
    }
````

````python
# [파이썬 코드 3-69]
def calc_metrics(prob, actual, threshold):
    pred = (prob > threshold).astype(int)
    cm = pd.crosstab(pred, actual).reindex(index=[0,1], columns=[0,1], fill_value=0)
    TN, FN = cm.loc[0,0], cm.loc[0,1]
    FP, TP = cm.loc[1,0], cm.loc[1,1]
    return {
        "임계값":   threshold,
        "정확도":    (TP+TN) / cm.values.sum(),
        "sensitivity": TP/(TP+FN) if (TP+FN) else np.nan,
        "specificity": TN/(TN+FP) if (TN+FP) else np.nan,
    }
thr_seq = np.arange(0.05, 0.96, 0.01)
trade_df = pd.DataFrame([calc_metrics(pred_prob, data["hp_int"], t) for t in thr_seq])
plt.figure(figsize=(8, 5))
plt.plot(trade_df["임계값"], trade_df["sensitivity"],
         color="#E07B91", linewidth=1.5, label="sensitivity")
plt.plot(trade_df["임계값"], trade_df["specificity"],
         color="#6888BE", linewidth=1.5, label="specificity")
plt.xlabel("임계값"); plt.ylabel("값")
plt.title("임계값에 따른 민감도와 특이도")
plt.legend(); plt.tight_layout(); plt.show()
````

````python
# [파이썬 코드 3-70]
X = data[["age", "bmi"]].values
y = data["hp_int"].values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=2025)
````

````python
# [파이썬 코드 3-71]
k_vals_cv = [1, 3, 5, 10, 15, 20, 30, 50, 100]
n_folds = 5
kf = KFold(n_splits=n_folds, shuffle=True, random_state=2025)
cv_err = []
for k in k_vals_cv:
    fold_err = []
    for train_idx, val_idx in kf.split(X_scaled):
        knn = KNeighborsClassifier(n_neighbors=k)
        knn.fit(X_scaled[train_idx], y[train_idx])
        fold_err.append(np.mean(knn.predict(X_scaled[val_idx]) != y[val_idx]))
    cv_err.append(np.mean(fold_err))
cv_results = pd.DataFrame({"K": k_vals_cv, "cv_error": cv_err})
cv_results
````

## A.4.1 확률과 확률분포

````python
# 성공 확률이 50%()인 공정한 동전을 10번 던지는 실험()을 생각해보자. 이때 앞면이 나오는 횟수를 확률변수 라고 할 때, 우리는 scipy.stats 라이브러리의 binom 모듈을 이용해 이론적인 확률을 계산할 수 있다.
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import binom, norm
# 설정: 시행 횟수 n=10, 성공 확률 p=0.5
n = 10
p = 0.5
# 1) P(X = 2): 앞면이 정확히 2회 나올 확률 (확률질량함수 pmf 사용)
prob_exact = binom.pmf(2, n=n, p=p)
print(f"P(X = 2) = {prob_exact:.4f}")
# 2) P(X <= 3): 앞면이 3회 이하로 나올 확률 (누적분포함수 cdf 사용)
prob_less_equal = binom.cdf(3, n=n, p=p)
print(f"P(X <= 3) = {prob_less_equal:.4f}")
# 3) P(X > 7): 앞면이 7회 초과로 나올 확률 (1 - cdf 사용)
prob_over = 1 - binom.cdf(7, n=n, p=p)
print(f"P(X > 7) = {prob_over:.4f}")
````

### 결과는 다음과 같다.

> 실행 대상이 아닌 설치 명령·오류 예시·출력 결과이다.

```text
P(X = 2) = 0.0439
P(X <= 3) = 0.1719
P(X > 7) = 0.0547
```

````python
# 다음으로 통계학의 꽃인 정규분포를 살펴보자. 평균이 0이고 표준편차가 1인 **표준정규분포()**에서 데이터를 무작위로 추출(Simulation)해 보자. 이론적으로 완벽한 곡선(PDF)과 무작위로 뽑힌 데이터의 분포(히스토그램)가 어떻게 다른지 비교하는 것이 핵심이다.
# 설정: 평균(mu)=0, 표준편차(sigma)=1
mu = 0
sigma = 1
size = 1000  # 표본 크기
# 데이터 생성 (시뮬레이션: 현실의 데이터)
np.random.seed(123)
simulated_data = np.random.normal(loc=mu, scale=sigma, size=size)
# 시각화: 경험적 분포(히스토그램) vs 이론적 분포(선 그래프)
plt.figure(figsize=(8, 5))
# 히스토그램 (경험적 확률)
plt.hist(simulated_data, bins=30, density=True, color='skyblue',
         edgecolor='black', alpha=0.7, label='Empirical (Data)')
# 이론적 밀도곡선
x = np.linspace(min(simulated_data), max(simulated_data), 1000)
pdf = norm.pdf(x, loc=mu, scale=sigma)
plt.plot(x, pdf, 'r-', label='Theoretical PDF')
plt.xlabel('X 값')
plt.ylabel('밀도')
plt.title(f'정규분포 시뮬레이션 비교 (mu={mu}, sigma={sigma})')
plt.legend()
plt.show()
````

````python
# 정규분포의 모양은 평균()과 표준편차()에 의해 결정된다. 이 파라미터들이 변할 때 분포가 어떻게 바뀌는지 확인해 보자. 다음 코드는 ,  각각 경우의 정규분포로부터 표집한 1000개의 표본을 보여준다.
# 시뮬레이션 설정
mu_list = [0, 10]
sigma_list = [1, 4]
sim_size = 1000
# 서브플롯 설정
fig, axes = plt.subplots(len(mu_list), len(sigma_list), figsize=(12, 8), constrained_layout=True)
# 시각화 반복
for i, mu_val in enumerate(mu_list):
    for j, sigma_val in enumerate(sigma_list):
        ax = axes[i, j]
        # 데이터 생성 및 히스토그램
        data = np.random.normal(loc=mu_val, scale=sigma_val, size=sim_size)
        ax.hist(data, bins=30, density=True, color='skyblue', edgecolor='black', alpha=0.7)
        # 이론적 곡선
        x_rng = np.linspace(mu_val - 4*sigma_val, mu_val + 4*sigma_val, 500)
        ax.plot(x_rng, norm.pdf(x_rng, mu_val, sigma_val), 'r-')
        ax.set_title(f"mu={mu_val}, sigma={sigma_val}")
plt.suptitle("파라미터에 따른 정규분포의 변화", fontsize=16)
plt.show()
````

````python
# 추론에서 자주 사용하게 될 중요한 확률 값들을 계산해 보자. 특히 1.96이라는 숫자에 주목하라.
mu = 0
sigma = 1
# 1) P(X <= 1.64): 상위 5% 지점 (단측 검정에서 주로 사용)
prob1 = norm.cdf(1.64, loc=mu, scale=sigma)
print(f"P(X <= 1.64) = {prob1:.4f}")
# 2) P(-1 < X < 1): 평균에서 1표준편차 이내에 있을 확률
prob2 = norm.cdf(1, loc=mu, scale=sigma) - norm.cdf(-1, loc=mu, scale=sigma)
print(f"P(-1 < X < 1) = {prob2:.4f}")
# 3) P(|X| >= 1.96): 양쪽 꼬리 확률의 합 (95% 신뢰구간의 근거)
prob3 = norm.cdf(-1.96, loc=mu, scale=sigma) + (1 - norm.cdf(1.96, loc=mu, scale=sigma))
print(f"P(|X| >= 1.96) = {prob3:.4f}")
````

````python
# 마지막으로, 확률이 주어졌을 때 그에 해당하는 값을 찾아보자. 이를 분위수라고 한다.
# 25%(Q1), 50%(중앙값), 75%(Q3) 분위수 구하기
quantiles = [0.25, 0.5, 0.75, 0.975]
q_values = norm.ppf(quantiles, loc=mu, scale=sigma)
for q, val in zip(quantiles, q_values):
    print(f"{int(q*100)}% 이론적 분위수: X = {val:.4f}")
````

````python
# 현실에서는 데이터를 한 번만 수집할 수 있지만, 컴퓨터 세상에서는 얼마든지 가상의 표집을 반복할 수 있다. 다음 함수는 모집단에서 지정된 크기(sample_size)만큼 표본을 뽑아 평균을 구하는 작업을 수천 번(num_simulations) 반복하여 리스트로 반환한다.
def simulate_sampling_distribution(population_df, sample_size, num_simulations):
    sample_means = []
    for _ in range(num_simulations):
        # 모집단에서 무작위 표본 추출 (복원 추출)
        sample = population_df['finish_net_minutes'].sample(n=sample_size, replace=True, random_state=None)
        # 표본 평균 계산 및 저장
        sample_mean = sample.mean()
        sample_means.append(sample_mean)
    return sample_means
````

## A.4.5 관악초등학교 학생의 평균 에너지 섭취량에 대한 신뢰구간

````python
# 우리는 관악초등학교 소득 상위 20% 가구 학생 10명을 무작위로 선정하여 조사했다. 하지만 현실 조사에서는 늘 예상치 못한 변수가 발생한다. 8번 학생이 응답을 거부하거나 기록이 누락된 것이다. 먼저 수집된 데이터를 입력하고, 분석 가능한 상태로 만들어보자. 결측값는 numpy 라이브러리의 nan(Not a Number)으로 표현한다.
import numpy as np
import scipy.stats as stats
# 1. 데이터 입력 (10명, 8번 학생은 결측)
# 결측값는 np.nan으로 표현
raw_data = np.array([1531, 1809, 2094, 1625, 1115,
                     1630, 1963, np.nan, 1744, 2485])
# 2. 전처리: 결측값 제거
# 통계 계산을 위해 결측값를 제외한 유효한 데이터만 남긴다.
data = raw_data[~np.isnan(raw_data)]
print(f"원래 표본 크기: {len(raw_data)}명")
print(f"유효 표본 크기(n): {len(data)}명")
print(f"데이터: {data}")
````

````python
# 이제 9명의 데이터를 바탕으로 표본평균()과 표본표준편차()를 계산한다. 여기서 주의할 점은 표준편차를 계산할 때 자유도를 고려하여 ddof=1 옵션을 사용해야 한다는 점이다.
# 3. 통계량 계산
n = len(data)
x_bar = np.mean(data)
s = np.std(data, ddof=1) # 표본표준편차 (자유도 n-1)
# 4. 표준오차(SE) 계산
# 모표준편차를 모르므로 표본표준편차(s)를 이용해 추정한다.
se = s / np.sqrt(n)
print("-" * 30)
print(f"표본 평균(x_bar): {x_bar:.2f} kcal")
print(f"표본 표준편차(s): {s:.2f} kcal")
print(f"표준오차(SE): {se:.2f} kcal")
print("-" * 30)
````

````python
# 자유도()를 고려한 t-값을 계산하고, 관악초등학교 학생의 평균 에너지 섭취량에 대한 신뢰구간을 구한다.
# 5. 신뢰구간
alpha = 0.05
df = n - 1
t_crit = stats.t.ppf(1 - alpha/2, df) # 자유도 8인 t-분포의 상위 2.5% 지점
margin_t = t_crit * se
lower_t = x_bar - margin_t
upper_t = x_bar + margin_t
print(f"\n자유도(df): {df}")
print(f"t-임계값: {t_crit:.4f}") # Z값 1.96보다 크다
print(f"[t-분포 사용 시] 95% 신뢰구간: [{lower_t:.2f}, {upper_t:.2f}]")
print(f"구간의 폭: {upper_t - lower_t:.2f}")
````

## 데이터 준비: 분석 대상 정의하기

````python
# 이번 실습에서는 국민건강영양조사(KNHANES) 2019년 데이터를 사용한다. 분석의 초점을 명확히 하기 위해, '20세 이하 남성' 그룹을 특정하여 그들의 에너지 섭취량(N_EN)을 분석 대상으로 삼았다.
import pandas as pd
import numpy as np
# 데이터 불러오기
hn19 = pd.read_csv("데이터/hn19.csv")
# 데이터 필터링: 에너지 섭취량이 기록된 20세 이하 남성
hn19_15 = hn19[
    hn19['N_EN'].notna() &
    (hn19['sex'] == 1) &
    (hn19['age'] <= 20)
].copy()
# 100명의 표본 추출 (랜덤 시드 고정)
sampled_data = hn19_15.sample(n=100, random_state=123, replace=False)
````

````python
# 이 과정을 수천 번 반복하여 중앙값들의 분포(부트스트랩 분포)를 만든다.
def bootstrap_median_ci(data, B=1999, conf=0.95):
    n = len(data)
    medians = []
    # 부트스트랩 반복: 가상의 표본을 만들고 중앙값을 기록한다
    for i in range(B):
        np.random.seed(i) # 재현성을 위해 시드 설정
        sample_i = np.random.choice(data, size=n, replace=True)
        medians.append(np.median(sample_i))
    # 퍼센타일(백분위수)을 이용한 신뢰구간 계산
    alpha = 1 - conf
    lower = np.percentile(medians, 100 * (alpha / 2))
    upper = np.percentile(medians, 100 * (1 - alpha / 2))
    estimate = np.median(data) # 원래 표본의 중앙값
    return {
        'estimate': float(round(estimate, 3)),
        'ci_lower': float(round(lower, 3)),
        'ci_upper': float(round(upper, 3))
    }
````

````python
# 이제 함수를 실행해 95% 신뢰구간을 구해보자. 반복 횟수 는 1,999번으로 설정한다. (원래 표본의 추정값 1개를 포함하여 총 2,000개를 맞추기 위함이다.)
# 'sampled_data'의 'N_EN' 컬럼(에너지 섭취량)에 대해 부트스트랩 수행
result = bootstrap_median_ci(sampled_data['N_EN'], B=1999, conf=0.95)
print(result)
{'estimate': 1824.3, 'ci_lower': 1698.687, 'ci_upper': 2010.36}
````

````python
# 먼저 분석에 필요한 데이터를 추출한다. 위의 예제에서는 표본 수 인 표본에 대한 예를 들었지만, 이 실습에서는 KNHANES 데이터의 모든 데이터를 전부 사용한다. 성별이 여성(sex == 2)이고 만 20세 이상인 성인(age >= 20) 중, 소득 수준이 하위 20%인 1분위(저소득층)와 상위 20%인 5분위(고소득층) 데이터를 선별한다.
import pandas as pd
import numpy as np
from scipy import stats
# 데이터 불러오기 (앞선 실습에서 로드된 hn19 데이터프레임 사용 가정)
# hn19 = pd.read_csv("hn19.csv")
# 1. 데이터 필터링 및 income_group 변수 생성
hn19_filtered = hn19[
    hn19['ho_incm5'].isin([1, 5]) &
    hn19['N_EN'].notna() &
    (hn19['sex'] == 2) &
    (hn19['age'] >= 20)
].copy()
# 가독성을 위해 그룹 이름 매핑
hn19_filtered['income_group'] = hn19_filtered['ho_incm5'].map({
    1: '저소득층',
    5: '고소득층'
}).astype('category')
# 2. 그룹별 요약 통계 계산
summary_stats = hn19_filtered.groupby('income_group')['N_EN'].agg(
    평균=('mean'),
    표준편차=('std'),
    n=('count')
).reset_index()
print("그룹별 칼로리 섭취량 요약:")
display(summary_stats)
````

````python
# 첫 번째 질문은 "저소득층 여성의 섭취량이 1,600kcal보다 유의하게 적은가?"이다. 우리가 입증하고 싶은 것은 "적다"이므로, alternative='less' 옵션을 사용하여 단측 검정을 수행한다.
# 저소득층 데이터 추출
low_income_data = hn19_filtered[hn19_filtered['income_group'] == '저소득층']['N_EN']
# 귀무가설의 모평균 값 (mu_0)
mu_0 = 1600
# 1. 일표본 t-검정 수행
# alternative='less': 대립가설이 '평균 < mu_0' 인 단측 검정
t_statistic_one_sample, p_value_one_sample = stats.ttest_1samp(
    low_income_data, popmean=mu_0, alternative='less'
)
print(f"저소득층 표본 크기: {len(low_income_data)}명")
print(f"저소득층 평균 칼로리 섭취량: {low_income_data.mean():.2f} kcal")
print("-" * 30)
print(f"일표본 t-통계량: {t_statistic_one_sample:.3f}")
print(f"일표본 P-value (단측, less): {p_value_one_sample:.3f}")
````

````python
# 두 번째 질문은 "저소득층과 고소득층 간에 섭취량 차이가 있는가?"이다. 앞서 기초 통계량에서 확인했듯, 두 집단의 표준편차(분산)가 꽤 다르다. 따라서 등분산을 가정하는 일반적인 t-검정 대신, 분산이 달라도 안전하게 사용할 수 있는 웰치의 t-검정(Welch's t-test)을 수행해야 한다. 파이썬에서는 equal_var=False 옵션으로 이를 간단히 적용할 수 있다.
# 고소득층 데이터 추출
high_income_data = hn19_filtered[hn19_filtered['income_group'] == '고소득층']['N_EN']
print(f"저소득층 표본 크기: {len(low_income_data)}명")
print(f"고소득층 표본 크기: {len(high_income_data)}명")
print("-" * 30)
# 2. 이표본 t-검정 (Welch's t-test)
# equal_var=False: 등분산 가정을 하지 않음 (Welch's t-test 수행)
t_statistic, p_value = stats.ttest_ind(
    low_income_data, high_income_data, equal_var=False
)
print(f"Welch's t-statistic: {t_statistic:.3f}")
print(f"P-value: {p_value:.3f}")
````

## 데이터 준비 및 탐색

````python
# 먼저 국민건강영양조사 데이터에서 성인 여성을 대상으로 소득 분위별(1~5분위)로 각각 100명씩을 무작위 추출하여 분석 데이터를 생성한다.
import pandas as pd
import numpy as np
# (데이터 로드 과정 생략: hn19 데이터프레임이 준비되었다고 가정)
# 1. 데이터 필터링: 성인 여성, 소득분위 1~5, 에너지 섭취량(N_EN) 있는 경우
data_anova = hn19[
    hn19['ho_incm5'].isin([1, 2, 3, 4, 5]) & # Changed to include all 5 income groups
    hn19['N_EN'].notna() &
    (hn19['sex'] == 2) &
    (hn19['age'] >= 19)
].copy()
# 2. 그룹별 100명 무작위 표집
sampled_data = data_anova.groupby('ho_incm5', group_keys=False).sample(n=100, replace=False, random_state=123)
# 3. 그룹별 기초 통계량 확인
summary = sampled_data.groupby('ho_incm5')['N_EN'].agg(['mean', 'std', 'count'])
print(summary)
````

## 분산분석(ANOVA) 수행

````python
# 분산분석을 위해 statsmodels 라이브러리를 사용한다. 여기서 핵심은 N_EN ~ C(ho_incm5)라는 수식이다. 이는 "에너지 섭취량(N_EN)을 소득 분위(ho_incm5)라는 범주형 변수(C)로 설명하겠다"는 의미다.
import statsmodels.api as sm
from statsmodels.formula.api import ols
# 4. ANOVA 모델 적합
# C(ho_incm5): 소득분위를 연속형 숫자가 아닌 범주형(Category) 변수로 인식하게 함
model = ols('N_EN ~ C(ho_incm5)', data=sampled_data).fit()
# 5. 분산분석표(ANOVA Table) 출력
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
````

## 튜키의 HSD 검정

````python
# ANOVA는 "차이가 있다"는 사실만 알려줄 뿐, 구체적으로 "1분위와 5분위가 다른지", "3분위와 4분위가 다른지"는 알려주지 않는다. 이를 확인하기 위해 Tukey의 HSD(Honest Significant Difference) 검정을 수행한다.
from statsmodels.stats.multicomp import pairwise_tukeyhsd
# 6. 사후 검정 (Post-hoc Test)
tukey = pairwise_tukeyhsd(endog=sampled_data['N_EN'],     # 데이터
                          groups=sampled_data['ho_incm5'], # 그룹
                          alpha=0.05)                      # 유의수준
print(tukey)
````

## 데이터 준비 및 교차표 생성

````python
# 먼저 분석 대상을 명확히 한다. 전체 데이터에서 '만 20세 이상 29세 이하의 성인 여성'만을 추출하고, 결측값가 있는 데이터는 제거한다. 분석할 변수는 소득 분위(ho_incm5)와 주관적 건강 상태(D_1_1)다.
import pandas as pd
from scipy.stats import chi2_contingency, permutation_test
import numpy as np
# 1. 데이터 필터링: 20대 여성 데이터 추출
# 결측값(9) 제거 및 필요한 변수만 선택
health_income_data = hn19[
    (hn19['sex'] == 2) &
    (hn19['age'] >= 20) &
    (hn19['age'] <= 29) &
    (hn19['D_1_1'] != 9)
][['D_1_1', 'ho_incm5']].dropna()
# 범주형 변수로 변환
health_income_data['D_1_1'] = health_income_data['D_1_1'].astype('category')
health_income_data['ho_incm5'] = health_income_data['ho_incm5'].astype('category')
print(f"분석 대상 데이터 크기: {len(health_income_data)}명")
# 2. 분할표 (Contingency Table) 생성
contingency_table = pd.crosstab(health_income_data['D_1_1'], health_income_data['ho_incm5'])
print("건강 상태(D_1_1)와 소득 분위(ho_incm5)의 교차표:")
display(contingency_table)
````

## 카이제곱 검정 수행

````python
# 대립가설  : 소득 분위와 주관적 건강 상태는 서로 독립이 아니다(관련이 있다).
# 3. 카이제곱 독립성 검정
chi2, p_value, dof, expected = chi2_contingency(contingency_table)
print(f"카이제곱 통계량: {chi2:.3f}")
print(f"P-value: {p_value:.3f}")
print(f"자유도 (Degrees of Freedom): {dof}")
print("기대 빈도표:\n", pd.DataFrame(expected, index=contingency_table.index, columns=contingency_table.columns))
````

````python
# 이때의 해결책 중 하나는 재범주화를 진행하는 것이다. 즉, 의미가 서로 다르지 않은 두개 이상의 범주를 하나로 합치는 방법으로 실전에서 제법 많이 쓰이는 방법이다. 예를 들어 건강 상태 1-2는 "좋음"으로, 3은 "보통"으로 4-5는 "나쁨"으로 재범주화한다면, 다음과 같이 셀의 값이 대략 5 이상인 분할표를 구할 수 있으며, 이 경우의 카이제곱 검정의 P-값은 정확하다고 볼 수 있다.
# D_1_1의 범주 재정의: (1,2) -> '좋음', 3 -> '보통', (4,5) -> '나쁨'
def recode_health_status(d11_value):
    if d11_value in [1, 2]:
        return '좋음'
    elif d11_value == 3:
        return '보통'
    elif d11_value in [4, 5]:
        return '나쁨'
    else:
        return np.nan # 정의되지 않은 값은 NaN으로 처리
health_income_data['health_status_recoded'] = health_income_data['D_1_1'].apply(recode_health_status).astype('category')
# ho_incm5는 기존대로 범주형 변수로 변환
health_income_data['ho_incm5'] = health_income_data['ho_incm5'].astype('category')
print(f"분석 대상 데이터 크기: {len(health_income_data)}명")
# 2. 분할표 (Contingency Table) 생성
contingency_table_3 = pd.crosstab(health_income_data['health_status_recoded'], health_income_data['ho_incm5'])
print("재범주화된 건강 상태와 소득 분위의 교차표:")
display(contingency_table_3)
````

````python
# 재범주화된 데이터를 이용하여 독립성 검정을 다시 시행한다.
chi2, p_value, dof, expected = chi2_contingency(contingency_table_3)
print(f"P-value: {p_value:.3f}")
````

## 최적 부분집합 선택법

````python
# [파이썬 코드 5-4]
# 전체 모형으로 오차분산(sigma^2) 추정
model_full = LinearRegression().fit(X, y)
rss_full = np.sum((y - model_full.predict(X)) ** 2)
sigma2_hat = rss_full / (n - p - 1)   # 불편추정량
tss = np.sum((y - y.mean()) ** 2)      # 총제곱합 (결정계수 계산용)
````

````python
# [파이썬 코드 5-5]
# 변수 개수별 최적 부분집합 탐색 및 BIC 계산
results = []
for k in range(1, p + 1):
    best_rss, best_vars = np.inf, None
    for cols in combinations(X.columns, k):
        rss = np.sum((y - LinearRegression().fit(X[list(cols)], y).predict(X[list(cols)])) ** 2)
        if rss < best_rss:
            best_rss, best_vars = rss, cols
    bic = n * np.log(best_rss / n) + k * np.log(n)
    results.append({"변수 개수": k, "선택 변수": best_vars, "BIC": bic})
reg_summary = pd.DataFrame(results).set_index("변수 개수")
print(reg_summary.round(3))
````

## 평가 기준을 통한 변수 개수 선택

````python
# [파이썬 코드 5-6]
# BIC 시각화
best_k = reg_summary["BIC"].idxmin()
plt.figure()
plt.plot(reg_summary.index, reg_summary["BIC"], marker="o")
plt.axvline(best_k, color="red", linestyle="--")
plt.scatter(best_k, reg_summary.loc[best_k, "BIC"], color="red", s=80)
plt.xlabel("변수 개수")
plt.ylabel("BIC")
plt.title("BIC 기준 최적 변수 개수 선택")
plt.show()
print(f"BIC 기준 최적 변수 개수: {best_k}개")
print(f"선택된 변수: {list(reg_summary.loc[best_k, '선택 변수'])}")
````

## 교차검증을 이용한 변수 개수 선택

````python
# [파이썬 코드 5-7]
# 10-겹 교차검증으로 변수 개수 선택
kf = KFold(n_splits=10, shuffle=True, random_state=1)
cv_errors = np.zeros((10, p))
for fold_idx, (train_idx, test_idx) in enumerate(kf.split(X)):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    for k in range(1, p + 1):
        best_rss, best_vars = np.inf, None
        for cols in combinations(X.columns, k):
            model = LinearRegression().fit(X_train[list(cols)], y_train)
            rss = np.sum((y_train - model.predict(X_train[list(cols)])) ** 2)
            if rss < best_rss:
                best_rss, best_vars = rss, cols
        model = LinearRegression().fit(X_train[list(best_vars)], y_train)
        cv_errors[fold_idx, k - 1] = mean_squared_error(
            y_test, model.predict(X_test[list(best_vars)])
        )
mean_cv = cv_errors.mean(axis=0)
optimal_k = np.argmin(mean_cv) + 1
print(optimal_k)
8
````

````python
# [파이썬 코드 5-41]
# K-평균 군집분석 (K=8)
k = 8
kmeans = KMeans(n_clusters=k, random_state=0, n_init=10)
cluster_labels = kmeans.fit_predict(subway_boarding_scaled)
print(cluster_labels[:10])
````

## 군집 결과 시각화

````python
# [파이썬 코드 5-42]
# PCA로 2차원 축소 후 군집 라벨 결합
pca = PCA(n_components=2)
subway_pca_2d = pca.fit_transform(subway_boarding_scaled)
subway_pc_df = pd.DataFrame(subway_pca_2d, columns=["PC1", "PC2"])
subway_pc_df["cluster"] = cluster_labels.astype(str)
# 산점도 시각화
plt.figure(figsize=(9, 6))
sns.scatterplot(data=subway_pc_df, x="PC1", y="PC2",
                hue="cluster", palette="tab10", alpha=0.7, s=20)
plt.title("K-평균 군집 결과 (PCA 2차원 시각화)")
plt.legend(title="군집", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
````
